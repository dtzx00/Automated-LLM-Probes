```python
import automated_llm_probes as alp

models = alp.ready_models()
models_to_try = [
    "Claude Haiku 4.5", "Claude Opus 4.5", "Claude Opus 4.7",
    "Claude Opus 5", "Claude Sonnet 4.5", "GPT-3.5-Turbo",
    "GPT-4-Turbo", "GPT-4o", "GPT-4o-mini", "GPT-5.4",
    "Grok 4.2", "Grok 4.3", "Grok 4.5", "Grok Build 0.1",
    "Llama-3.1 8b", "Llama-3.2 3b", "Llama-4 Maverick", "Llama-4 Scout",
]
cues_to_try = [
    "superpower",
    "title (2305 or Execution)",
    "frame",
    "glow",
    "death",
    "delay",
    "enemy",
    "illness",
    "lie",
    "marriage",
    "joy",
    "shade",
    "simplicity",
    "sky",
]
models_to_try = [m for m in models if m["name"] in models_to_try]
print(sorted(m["name"] for m in models_to_try))

ADD = 30

def n_have(model):
    mdir = alp._model_dir("CWT", model["name"], 0.5)
    return sum(1 for p in mdir.glob("*.pickle") if p.is_file())

print("current pickle totals")
for m in models_to_try:
    print(f"  {m['name']:20s} {n_have(m)}")

for cue in cues_to_try:
    print("\n===", cue, "===")
    for m in models_to_try:
        have = n_have(m)
        print(f"  {m['name']}: {have} -> {have + ADD}")
        alp.collect("CWT", models=[m], n_per_model=have + ADD, cue=cue)
```
