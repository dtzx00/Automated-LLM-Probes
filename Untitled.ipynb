```python
import os, pickle
import automated_intelligence_tests as ait
from tqdm import tqdm

def list_pickle_fps(root):
    fps = []
    for dp, _, fns in os.walk(root):
        for n in fns:
            if n.endswith(".pickle") and not n.endswith(".pickle.tmp"):
                fps.append(os.path.join(dp, n))
    fps.sort()
    return fps

def dump(p, row):
    tmp = p + ".tmp"
    with open(tmp, "wb") as f:
        pickle.dump(row, f, protocol=pickle.HIGHEST_PROTOCOL)
    os.replace(tmp, p)
```

```python
def _score_row(row):
    """Write row['parsed'] and row['score']. Never delete the pickle."""
    task = (row.get("task") or "").strip().lower()
    raw = row.get("raw")
    stim = row.get("kwargs") or None

    if row.get("error") or not raw or task not in TASKS:
        row["parsed"] = None
        row["score"] = None
        return row
    if task == "cwt" and _cwt_cue_poisoned(stim):
        row["parsed"] = None
        row["score"] = None
        return row

    try:
        parsed = ait.parse(task, raw, stim=stim)
    except Exception:
        parsed = None
    row["parsed"] = parsed

    if parsed is None:
        row["score"] = None
        return row

    try:
        result = ait.evaluate(task, parsed)
        score = None if result is None else result.get("score")
    except Exception:
        score = None
    row["score"] = score if score is None or isinstance(score, (int, float)) else None
    return row


def backfill_task(task, overwrite=OVERWRITE):
    task = task.strip().lower()
    fps = sorted(p for p in (DATA_ROOT / task).rglob("*.pickle") if p.is_file())
    n_ok = n_none = n_skip = n_bad = 0
    for fp in tqdm(fps, desc=task):
        try:
            with fp.open("rb") as f:
                row = pickle.load(f)
        except Exception as e:
            print(f"SKIP {fp}: {e}")
            n_bad += 1
            continue
        if not overwrite and isinstance(row.get("score"), (int, float)):
            n_skip += 1
            continue
        try:
            _score_row(row)
            _atomic_dump(fp, row)
        except Exception as e:
            print(f"FAIL {fp}: {e}")
            n_bad += 1
            continue
        if row.get("score") is None:
            n_none += 1
        else:
            n_ok += 1
    print(
        f"{task}: scored={n_ok} score=None={n_none} "
        f"skipped={n_skip} bad={n_bad} total={len(fps)}"
    )
```
